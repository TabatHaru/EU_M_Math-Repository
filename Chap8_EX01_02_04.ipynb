```python
import requests, zipfile
import io
import pandas as pd

# Corrected URL
url = 'https://archive.ics.uci.edu/ml/machine-learning-databases/autos/imports-85.data'
res = requests.get(url).content

# Use comma as the delimiter (the data is comma-separated)
auto = pd.read_csv(io.StringIO(res.decode('utf-8')), header=None, sep=',')

auto.columns = ['symboling', 'normalized-losses','make','fuel-type','aspiration','num-of-doors',
                'body-style','drive-wheels','engine-location','wheel-base','length','width','height',
                'curb-weight','engine-type','num-of-cylinders','engine-size','fuel-system','bore',
                'stroke','compression-ratio','horsepower','peak-rpm','city-mpg','highway-mpg','price']

print('自動車データの形式 : {}'.format(auto.shape))

auto.head()
```

```python
auto = auto[['price','horsepower','width','height']]
auto.isin(['?']).sum()
```

```python
auto = auto.replace('?', np.nan).dropna()
print('自動車データの形式 : {}'.format(auto.shape))
```

```python
print('データの型の確認 (型変換前) \n{}\n'.format(auto.dtypes))
```

```python
auto = auto.assign(price=pd.to_numeric(auto['price']))
auto = auto.assign(horsepower=pd.to_numeric(auto['horsepower']))
print()
print('データの型の確認 (型変換後) \n{}\n'.format(auto.dtypes))
```

```python
auto.corr()
```

```python
from sklearn.model_selection import train_test_split

from sklearn.linear_model import LinearRegression

x = auto.drop('price', axis=1)
y = auto['price']

x_train, x_test, y_train, y_test = train_test_split(x, y, test_size=0.3, random_state=0)

model = LinearRegression()
model.fit(x_train, y_train)

print('決定係数(train):(:.3f)'.format(model.score(x_train, y_train)))
print('決定係数(test):(:.3f)'.format(model.score(x_test, y_test)))

print('回帰係数 \n{}'.format(pd.Series(model.coef_, index=x.columns)))
print('切片 \n{}'.format(model.intercept_))
```

```python
# Practice 8-1
import requests, zipfile
import io

url = 'https://archive.ics.uci.edu/ml/machine-learning-databases/autos/imports-85.data'
res = requests.get(url).content

auto = pd.read_csv(io.StringIO(res.decode('utf-8')), header=None, sep=',')
auto.columns = ['symboling', 'normalized-losses','make','fuel-type','aspiration','num-of-doors',
                'body-style','drive-wheels','engine-location','wheel-base','length','width','height',
                'curb-weight','engine-type','num-of-cylinders','engine-size','fuel-system','bore',
                'stroke','compression-ratio','horsepower','peak-rpm','city-mpg','highway-mpg','price']

from sklearn.model_selection import train_test_split

from sklearn.linear_model import LinearRegression

auto = auto[['price','horsepower','width','height']]
auto = auto.replace('?', np.nan).dropna()

X = auto.drop('price', axis=1)
y = auto['price']

X_train, X_test, y_train, y_test = train_test_split(X, y, test_size=0.3, random_state=0)

model = LinearRegression()
model.fit(X_train, y_train)

print('決定係数(train):{:.3f}'.format(model.score(X_train, y_train)))
print('決定係数(test):{:.3f}'.format(model.score(X_test, y_test)))
```

```python
# Practice 8-2
from sklearn.datasets import load_breast_cancer
from sklearn.preprocessing import StandardScaler
from sklearn.linear_model import LogisticRegression

cancer = load_breast_cancer()

X_train, X_test, y_train, y_test = train_test_split(cancer.data, cancer.target, test_size=0.3, random_state=0)

model = LogisticRegression()
model.fit(X_train, y_train)

print('決定係数(train):{:.3f}'.format(model.score(X_train, y_train)))
print('決定係数(test):{:.3f}'.format(model.score(X_test, y_test)))
```

```python
# Practice 8-4
from sklearn.linear_model import LinearRegression, Lasso

X = auto.drop('price', axis=1)
y = auto['price']

X_train, X_test, y_train, y_test = train_test_split(X, y, test_size=0.3, random_state=0)

models = {
    'linear': LinearRegression(),
    'lasso1': Lasso(alpha=1.0, random_state=0),
    'lasso2': Lasso(alpha=200.0, random_state=0)
}

score = {}

for model_name, model in models.items():
    model.fit(X_train, y_train)
    score[model_name, 'test'] = model.score(X_test, y_test)
    score[model_name, 'train'] = model.score(X_train, y_train)

pd.Series(score).unstack()
```

**ラッソとリッジ回帰で用いられる正則化項式のメリット**

係数の縮小：正則化により、全ての係数が少しずつ縮小されることにより、モデルの複雑さが抑えられ、過学習を防ぐ。

安定性の向上：データに対するロバスト性が向上し、データに含まれるノイズや多重共線性がある場合でも、より安定した予測結果が得られる。

モデルの簡素化：不要な変数が削除されるため、より簡素で解釈しやすいモデルが得られる。
